```python
import pandas as pd
import numpy as np


df = pd.read_excel(r"C:\Users\pns91\Downloads\Customer Churn  DAP\Telco_customer_churn.xlsx")


```

```python
df.head()
```

```python
df.columns = df.columns.str.lower()
```

```python
df.columns = df.columns.str.replace(' ', '_')
```

```python
print("New columns:", df.columns.tolist())
```

```python
df['total_charges'] = df['total_charges'].replace(' ', np.nan)
```

```python
df['total_charges'] = pd.to_numeric(df['total_charges'])
```

```python
df = df.dropna(subset=['total_charges'])
```

```python
df['churn_reason'] = df['churn_reason'].fillna('No Churn')
```

```python
print("Missing values left:\n", df.isnull().sum())
```

```python
columns_to_drop = ['count', 'country', 'state', 'churn_label']
```

```python
df = df.drop(columns=columns_to_drop)
```

```python
df = df.rename(columns={'churn_value': 'churn'})
```

```python
df.to_csv("cleaned_telco_churn.csv", index=False)

print("Data cleaning complete! Final size:", df.shape)
```
